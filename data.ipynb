```python
if patients is not None and 'age_as_of_2026' in patients.columns:     
    plt.figure(figsize=(8,5))     
    plt.hist(patients['age_as_of_2026'].dropna(), bins=25)     
    plt.xlabel('Age (as of 2026)')     
    plt.ylabel('Number of Patients')     
    plt.title('Age Distribution')     
    plt.show()
```

```python
if patients is not None and 'gender' in patients.columns:
    patients['gender'].value_counts(dropna=False).plot(kind='bar', figsize=(7,5))
    plt.title('Gender')
    plt.ylabel('count')
    plt.show()
```

```python
if outcomes is not None and 'lost_to_followup_label' in outcomes.columns:
    outcomes['lost_to_followup_label'].value_counts().plot(kind='bar', figsize=(7,5))
    plt.title('Lost to follow-up')
    plt.xlabel('0 = no, 1 = yes')
    plt.ylabel('episodes')
    plt.show()
```

```python
if outcomes is not None and 'dropout_stage_label' in outcomes.columns:
    outcomes['dropout_stage_label'].value_counts(dropna=False).plot(kind='barh', figsize=(9,5))
    plt.title('LTFU')
    plt.show()
```

```python
if tele is not None and 'distance_to_facility_km' in tele.columns:
    plt.figure(figsize=(8,5))
    plt.hist(tele['distance_to_facility_km'].dropna(), bins=30)
    plt.xlabel('distance to facility (km)')
    plt.ylabel('consultations')
    plt.title('Distance to facility')
    plt.show()
```

```python
if tele is not None and 'connectivity_quality' in tele.columns:
    tele['connectivity_quality'].value_counts(dropna=False).plot(kind='bar', figsize=(8,5))
    plt.title('Connectivity quality')
    plt.ylabel('consultations')
    plt.show()
```

```python
if tele is not None and 'consult_date' in tele.columns:
    tele['consult_date'] = pd.to_datetime(tele['consult_date'], errors='coerce')
    monthly = tele.set_index('consult_date').resample('ME').size()

    plt.figure(figsize=(10,5))
    plt.plot(monthly.index, monthly.values, marker='o')
    plt.title('Teleconsultations over time')
    plt.ylabel('consultations')
    plt.xlabel('month')
    plt.xticks(rotation=45)
    plt.show()
```

```python
if tele is not None and outcomes is not None:
    x = tele.merge(outcomes[['teleconsult_id','lost_to_followup_label']], on='teleconsult_id', how='left')
    x.head()
```

```python
if 'x' in globals() and 'connectivity_quality' in x.columns:
    print(pd.crosstab(x['connectivity_quality'], x['lost_to_followup_label'], normalize='index').round(3))
            
    pd.crosstab(x['connectivity_quality'], x['lost_to_followup_label'], normalize='index').plot(kind='bar', figsize=(8,5))
    plt.title('LTFU by connectivity quality')
    plt.ylabel('proportion')
    plt.show()
```

```python
if 'x' in globals() and 'distance_to_facility_km' in x.columns:
    x['distance_group'] = pd.cut(x['distance_to_facility_km'], [0,5,10,20,30,50,100], include_lowest=True)
    print(pd.crosstab(x['distance_group'], x['lost_to_followup_label'], normalize='index').round(3))
```

```python
if 'x' in globals() and 'distance_group' in x.columns:
    x.groupby('distance_group')['lost_to_followup_label'].mean().plot(kind='bar', figsize=(8,5))
    plt.title('LTFU rate by distance')
    plt.ylabel('LTFU rate')
    plt.xlabel('distance group')
    plt.show()
```

```python
if 'x' in globals() and 'age' in x.columns:
    x['age_group'] = pd.cut(x['age'], [0,20,30,40,50,60,70,80,100])
    print(x.groupby('age_group', observed=True)['lost_to_followup_label'].agg(['mean','count']))
```

```python
if 'x' in globals() and 'age_group' in x.columns:
    x.groupby('age_group', observed=True)['lost_to_followup_label'].mean().plot(kind='bar', figsize=(9,5))
    plt.title('LTFU rate by age group')
    plt.ylabel('LTFU rate')
    plt.show()
```
